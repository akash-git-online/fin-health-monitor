```python

%sql
--Create catalog
CREATE CATALOG IF NOT EXISTS fhm_dev;

```

```python
%sql
--Create Schemas
CREATE SCHEMA  IF NOT EXISTS fhm_dev.meta;
CREATE SCHEMA  IF NOT EXISTS fhm_dev.landing;
CREATE SCHEMA IF NOT EXISTS fhm_dev.bronze;
CREATE SCHEMA IF NOT EXISTS fhm_dev.silver;
CREATE SCHEMA IF NOT EXISTS fhm_dev.gold;
```

```python
%sql
COMMENT ON SCHEMA fhm_dev.landing IS 'Layer 0. Raw files exactly as received from sources (JSON, CSV), stored in Volumes. Never edited. One folder per source and run date.';

COMMENT ON SCHEMA fhm_dev.bronze IS 'Layer 1. Raw data loaded into Delta tables as received, plus ingestion metadata (source file, load time). No business logic.';

COMMENT ON SCHEMA fhm_dev.silver IS 'Layer 2. Cleaned, typed and conformed data. Duplicates resolved, restated filings kept as history. Source of truth for analysis.';

COMMENT ON SCHEMA fhm_dev.gold IS 'Layer 3. Business-ready tables and metrics (ratios, health status) used by dashboards, the app and Genie.';

COMMENT ON SCHEMA fhm_dev.meta IS 'Pipeline control data, not business data: source config, ingestion run log, data quality rules, metric definitions.';
```

```python
%sql
--Create Volume
CREATE VOLUME  IF NOT EXISTS fhm_dev.landing.raw;
```
